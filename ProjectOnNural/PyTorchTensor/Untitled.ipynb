```python
import gradio as gr
import time

def mFun(range_value):
    """Print values from 0 to range_value (exclusive)."""
    output = []

    for i in range(int(range_value)):
        output.append(str(i))
        time.sleep(0.1)

    return "\n".join(output)


def run_function(range_value):
    """Call mFun and return results."""
    return mFun(range_value)


# Create Gradio interface
with gr.Blocks(title="Range Value Printer") as demo:

    gr.Markdown("# Range Value Printer")
    gr.Markdown(
        "Adjust the range value and click Run to see numbers "
        "from 0 to the given range."
    )

    with gr.Row():

        # Left column
        with gr.Column(scale=1):
            range_slider = gr.Slider(
                minimum=1,
                maximum=20,
                value=10,
                step=1,
                label="Range Value"
            )

            run_button = gr.Button("Run", variant="primary")

        # Right column
        with gr.Column(scale=2):
            output_text = gr.Textbox(
                label="Output (0 to range-1)",
                lines=10,
                interactive=False,
                placeholder="Click Run to see the numbers..."
            )

    # Connect button click inside the Blocks context
    run_button.click(
        fn=run_function,
        inputs=[range_slider],
        outputs=[output_text]
    )


# Launch the app
demo.launch()
```

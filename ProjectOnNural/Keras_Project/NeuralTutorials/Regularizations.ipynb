#### L1 Regularization: a deep dive

L1 regularization adds a penalty term to the model's loss function, which is the sum of the absolute values of the coefficients multiplied by a regularization parameter (lambda, λ).

This penalty encourages the model to shrink the coefficients of irrelevant or less important features towards zero, effectively eliminating them from the model. This process is akin to automatic feature selection.

# Key characteristics and benefits

**Feature Selection**: L1 regularization automatically selects the most relevant features by forcing the coefficients of less significant features to become zero.


#### L2 regularization: a deep dive

L2 regularization, also known as **Ridge Regression**, is a common technique in machine learning that addresses the problem of overfitting.

Reduces Overfitting: By penalizing large coefficients, L2 regularization helps prevent the model from becoming too complex and overfitting the training data, leading to better generalization on new, unseen data.

#### Layer weight regularizers keras

**kernel_regularizer**: Regularizer to apply a penalty on the layer's kernel
**bias_regularizer**: Regularizer to apply a penalty on the layer's bias
**activity_regularizer**: Regularizer to apply a penalty on the layer's output

##### L1 class

The L1 regularization penalty is computed as:

loss = l1 * reduce_sum(abs(x))

##### L2 class:

The L2 regularization penalty is computed as: loss = l2 * reduce_sum(square(x))

##### L1L2 class:

A regularizer that applies both L1 and L2 regularization penalties.

The L1 regularization penalty is computed as: loss = l1 * reduce_sum(abs(x))

The L2 regularization penalty is computed as loss = l2 * reduce_sum(square(x))

```python

reg_val = 0.01
layer = layers.Dense(units=5,
                     kernel_initializer='ones',
                     kernel_regularizer=regularizers.L1(reg_val),
                     activity_regularizer=regularizers.L2(reg_val)
                     )
tensor = ops.ones(shape=(5, 5)) * 2.0
out = layer(tensor)
# The kernel regularization term is 0.25
# The activity regularization term (after dividing by the batch size) is 5
print('Output after regularions:',ops.sum(layer.losses).numpy())
print(layer.losses[0],layer.losses[1])

print('losses multiply by: ',layer.losses[0].numpy() + tf.size(tensor).numpy()* reg_val)
```

```python
# trensor array with 5x2 filled with random integer value [0,10]
X = tf.random.uniform(shape=(5, 2), minval=0, maxval=10, dtype=tf.int32)
print(X.numpy(),X.shape)
# create tensror array with shape (f,u) filled with random integer value 1
def create_numpy_array(f, u):
    """Create a numpy array of shape (f, u) filled with ones."""
    return np.ones((f, u), dtype=np.float32)*3.0
```

```python
IN = Input(shape=(X.shape[1],))
f = X.shape[1]  # number of features
u = 3 # number of neurons in the layer
# kernal weight initilazion by 1
# weight shape = (f,u), where f = number of feature, u = number of neuron
#

arr = create_numpy_array(f, u)
val = initializers.Constant(value=arr)
VL = val(shape=(f, u))
print('weight matrix :',VL.numpy())
#initilazion of baise, 
# Bias is the linear array of shape (u,)
# where u = number of neuron in the layer
reg_val = 0.02
L1 = Dense(units = u,
                    kernel_initializer=VL,
                    kernel_regularizer=regularizers.L1(reg_val),
                    activation = 'sigmoid',
                    name='dense_1')

out = L1(X)
w,b = L1.get_weights()
#print('Weights:', w)
# The kernel regularization term is 0.25
# The activity regularization term (after dividing by the batch size) is 5
print('Output after regularions:',ops.sum(L1.losses).numpy())
```

### The Perceptron

The Perceptron is one of the simplest ANN architectures. It is based on a slightly different artificial neuron
called a threshold logic unit (TLU), or sometimes a linear threshold unit (LTU).


#### TLU (threshold logic unit) or LTU (linear threshold unit): 

the ***inputs and output are now numbers*** (instead of binary on/off values) and each input connection
is associated with a weight. 

The TLU computes a weighted sum of its inputs 
z = w1 x1 + w2 x2 + ⋯ + wn xn = x^T w

then applies a step function to that sum and outputs the result: 

hw(x) = step(z), where z = x^T w.






A single TLU can be used for simple linear binary classification. you could use a single TLU to classify iris flowers based on the petal length and width (also adding an extra bias feature x0 = 1, just like we did in
previous chapters).

#### fully connected layer or a dense layer:

A Perceptron is simply composed of a single layer of TLUs,6 with each TLU connected to all the inputs. When all the neurons in a layer are connected to every neuron in the previous layer (i.e., its input neurons), it is called a fully connected layer or a dense layer.

Perceptron diagram:



#### Equation Computing the outputs of a fully connected layer:




- ***X*** represents the matrix of input features. It has one row per instance, one column per feature.
- The ***weight matrix W*** contains all the connection weights except for the ones from the bias neuron. It  
  has one row per input neuron and one column per artificial neuron in the layer.
- The ***bias vector b*** contains all the connection weights between the bias neuron and the artificial 
  neurons. It has one bias term per artificial neuron.
- ***The function ϕ is called the activation function***: when the artificial neurons are TLUs, it is a step 
  function (but we will discuss other activation functions shortly).

### Regression MLPs

MLPs can be used for regression tasks.

If you want to predict a single value (e.g., the price of a house given many of its features), then you just need a single output neuron: its output is the predicted value.

##### Qus: How many nuron required to locate center of an object on an image and building box around object?

Ans: 4

To locate the center of an object on an image, you need to predict 2D coordinates, so you need two output neurons. If you also want to place a bounding box around the object, then you need two more numbers: the width and the height of the object. So you end up with 4 output neurons.

##### Qus: Which activations functions usen in regressions MLP?

Do not want to use any activation function:

In general, when building an MLP for regression, you do not want to use any activation function for the output neurons, so they are free to output any range of values.

softplus activation function:

if you want to guarantee that the output will always be positive, then you can use the ReLU activation function or softplus activation function

logistic function or the hyperbolic tangent:

if you want to guarantee that the predictions will fall within a given range of values, then you can use the logistic function (0 to 1 for the logistic function) or the hyperbolic tangent (–1 to 1 for
the hyperbolic tangent).

##### Losses functions use in Regression MLPs:

loss function to use during training is typically the mean squared error, but if you have a lot of outliers in the training set, you may prefer to use the mean absolute error i.e.use the Huber loss.








#### Classification MLPs

Used for classification tasks.

***For a binary classification problem, you just need a single output neuron using the logistic activation function:***

The output will be a number between 0 and 1, which you can interpret as the estimated probability of the positive class.

***MLPs can also easily handle multilabel binary classification tasks:***

For example, you could have an email classification system, that predicts whether each incoming email is ham or spam and simultaneously predicts whether it is an urgent or non-urgent email.

Total output nuron required : 2

both using the logistic activation function: the first would output the probability that the email is
spam and the second would output the probability that it is urgent.

Another example: classes 0 through 9 for digit image classification, hen you need to have one output neuron per class i.e. 10

Should use the softmax activation function, the estimated probabilities are between 0 and 1.

### 3. Metrics

A metric is a function that is used to judge the performance of your model.

Metric functions are similar to loss functions, except that the results from evaluating a metric are not used when training the model. Note that you may use any loss function as a metric.

Type of Metrice:

- Base Metric class
- Accuracy metrics
- Probabilistic metrics
- Regression metrics
- Classification metrics based on True/False positives & negatives
- Image segmentation metrics
- Hinge metrics for "maximum-margin" classification
- Metric wrappers and reduction metrics

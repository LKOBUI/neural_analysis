### Sequential Neural Networks

A sequential neural network follows a layer-by-layer structure, where data flows in one direction—from the input layer to the output layer without skipping any layers. This structure is commonly used for feedforward networks.

Here, each layer is stacked sequentially—meaning there are no branching connections.

### Non-Sequential Neural Networks (Functional API)

Unlike sequential models, non-sequential (functional) networks allow:

- Multiple input/output layers
- Skip connections (like in ResNet)
- Dynamic architectures such as GANs or attention mechanisms

This model has multiple outputs—something not possible with Sequential.

***This architecture makes it possible for the neural network to learn both deep patterns (using the deep path) and simple rules (through the short path).***

### Faster Optimizers:

#### Momentum Optimization

Qus: Problem with Gradient Descent ?

Recall that Gradient Descent simply updates the weights $θ$ by directly subtracting the gradient of the cost function $J(θ)$ with regards to the weights $(∇θJ(θ))$ multiplied by the learning rate $η$.
The equation is: 

$θ ← θ – η∇θJ(θ)$

It does not care about what the earlier gradients were. If the local gradient is tiny, it goes very slowly.

In Momentum optimization cares a great deal about what previous gradients were the gradient is used for acceleration, not for speed. To simulate some sort of friction mechanism and prevent the momentum from growing too large, the algorithm introduces a new hyperparameter $β$, simply called the momentum, which must be set between $0$ (high friction) and $1$ (no friction).

Momentum algorithm



#### Explain Momentum algorithm (how work)

![alt text](image-1.png)

You can easily verify that if the gradient remains constant, the terminal velocity (i.e.,the maximum size of the weight updates) is equal to that gradient multiplied by the learning rate η multiplied by $1/(1 − β)$ (ignoring the sign).

For example, if β = 0.9, then the terminal velocity is equal to 10 times the gradient times the learning rate, so Momentum optimization ends up going 10 times faster than Gradient Descent! This allows Momentum optimization to escape from plateaus much faster than Gradient Descent.

##### Learning Rate Scheduling

- If you set it way too high, training may actually diverge it will make progress very quickly at first, but it 
  will end up dancing around the optimum, never really settling down.

- If you set it too low, training will eventually converge to the optimum, but it will take a very long time.

**Learning schedules:**

if you start with a high learning rate and then reduce it once it stops making fast progress, you can reach a
good solution faster than with the optimal constant learning rate.


Learning schedules strategies:

***Power scheduling:[ Power Scheduling (a.k.a. Polynomial Decay) learning rate]*** 

**Qus: What is Power Scheduling?**

Power scheduling is a learning rate decay strategy where the **learning rate decreases as a polynomial function** of the training step.
#### Mathematical Formulation
Let:
- $\eta_t$: learning rate at step $t$
- $\eta_0$: initial learning rate
- $t$: current training step (iteration or epoch)
- $T$: total number of steps (or decay horizon)
- $\alpha$ : power (decay exponent), typically 0<\alpha \leq 1



Intuition
- At the beginning ($t=0$):
  - $\eta _t$ = $\eta _0$ $→$ full learning rate.

- As t $\rightarrow$ T:
  - $\eta _t$ $→$ 0 → learning rate shrinks to zero.

- The exponent $\alpha$  controls the decay speed:
  - If $\alpha$ $= 1$: linear decay.
  - If $\alpha$ $< 1$: slower decay (flatter curve).
  - If $\alpha$ $> 1$: faster decay (steeper curve).

```python
# Power rule decay learning rate schedule visualization
# Parameters
eta0_values = [0.1, 0.3, 0.5, 0.7, 0.9, 1.0]   # initial learning rates
T = 100                                        # total steps

# Steps
t = np.arange(0, T+1)

plt.figure(figsize=(10,6))

# Loop over each eta0 and corresponding alpha
for eta0 in eta0_values:
    alpha = eta0 * 5
    eta_t = eta0 * (1 - t/T)**alpha
    plt.plot(t, eta_t, linewidth=2,
             label=fr'$\eta_0={eta0}, \alpha={alpha}$')

    # Highlight example points
    example_steps = [0, 25, 50, 100]
    plt.scatter(example_steps,
                eta0 * (1 - np.array(example_steps)/T)**alpha,
                s=40, zorder=5)

# Plot formatting
plt.title("Polynomial Decay Learning Rate (Multiple η₀ values)")
plt.xlabel("Training Step (t)")
plt.ylabel(f"Learning Rate ($\eta_t$)")
plt.legend()
plt.grid(True)
plt.show()
```



- Higher $\eta _0$ $→$ starts training with a larger step size.
- Larger $\alpha$  (since tied to $\eta _0$) → accelerates the decay, meaning the learning rate drops faster.
- Lower $\eta _0$ values **keep the learning rate small but more consistent across training**.

#### Qus: Why It Matters

- Early exploration vs. later fine-tuning:
- Large $\eta _0$ with steep decay → aggressive exploration at the start, then quickly stabilizes.
- Small $\eta _0$ with gentle decay → conservative updates, steady progress.

#### Key Observations from the Annotated Point

- **At step 0**: All curves start at their respective $\eta _0$ values (0.1 $→$ 1.0).

- **At step 25**:
- Smaller $\eta _0$ (e.g., 0.1, 0.3) decays gently, values remain close to the starting point.
- Larger $\eta _0$ (e.g., 0.9, 1.0) already show significant reduction due to higher $\alpha$ .

- **At step 50**:
- Curves with higher $\alpha$  (brown, purple) drop sharply, learning rate is near zero.
- Lower $\alpha$  curves (blue, orange) still retain a meaningful learning rate.
- At step 100: All curves converge to 0, since $(1-t/T)=0$.

#### Exponential scheduling:

Exponential Learning Rate Scheduling:

Exponential scheduling reduces the learning rate smoothly over time according to:



As t increases, the exponential term  shrinks, so the learning rate decays continuously.


```python
# Exponential decay learning rate schedule visualization
# Parameters
eta0 = 0.1                 # initial learning rate
lambda_values = [0.01, 0.03, 0.05, 0.07, 0.1]   # decay rates
T = 100                    # total steps

# Steps
t = np.arange(0, T+1)

plt.figure(figsize=(10,6))

# Loop over each lambda
for lambda_ in lambda_values:
    eta_t = eta0 * np.exp(-lambda_ * t)
    plt.plot(t, eta_t, linewidth=2,
             label=fr'$\lambda={lambda_}$')

    # Highlight example points
    example_steps = [0, 25, 50, 100]
    plt.scatter(example_steps,
                eta0 * np.exp(-lambda_ * np.array(example_steps)),
                s=40, zorder=5)

# Plot formatting
plt.title("Exponential Decay Learning Rate (Multiple λ values)")
plt.xlabel("Training Step (t)")
plt.ylabel("Learning Rate (η_t)")
plt.legend()
plt.grid(True)
plt.show()
```

#### Early learning vs. late fine-tuning:

- Small $\lambda$  (0.01): Slow decay; retains high learning rates, favoring exploration but risking sluggish convergence.
- Large $\lambda$  (0.1): Rapid decay; stabilizes quickly, but may underfit if the model needs more learning capacity early.
- Mid-range $\lambda$  (0.03–0.05): Smooth reduction; meaningful learning early with a clear shift to fine-tuning by mid–late training




#### Piecewise constant scheduling

Piecewise constant scheduling keeps the learning rate fixed for certain intervals, then drops it abruptly at predefined boundaries.





```python
# Piecewise constant learning rate schedule visualization
# Parameters
T = 100
boundaries = [10, 20, 30, 40, 50, 60, 70, 80, 90]   # step boundaries
values = [0.1, 0.08, 0.06, 0.05, 0.04, 0.03, 0.02, 0.01, 0.005, 0.001]  # learning rates

# Steps
t = np.arange(0, T+1)
eta_t = np.zeros_like(t, dtype=float)

# Assign learning rates piecewise
for i in range(len(boundaries)+1):
    if i == 0:
        eta_t[:boundaries[i]] = values[i]
    elif i < len(boundaries):
        eta_t[boundaries[i-1]:boundaries[i]] = values[i]
    else:
        eta_t[boundaries[-1]:] = values[i]

# Plot
plt.figure(figsize=(10,6))
plt.step(t, eta_t, where='post', linewidth=2, color='blue',
         label="Piecewise Constant Schedule")

# Annotate learning rate values at boundaries
for i, b in enumerate(boundaries + [T]):
    plt.text(b, values[i], f"{values[i]:.3f}", 
             ha='left', va='bottom', fontsize=9, color='red')

plt.title("Piecewise Constant Learning Rate Scheduling (Annotated)")
plt.xlabel("Training Step (t)")
plt.ylabel("Learning Rate (η_t)")
plt.legend()
plt.grid(True)
plt.show()
```

- Early training (0–30 steps): High learning rates (0.1 → 0.05) encourage faster exploration and bigger updates.
- Mid training (40–70 steps): Moderate rates (0.04 → 0.02) stabilize learning, reducing oscillations.
- Late training (80–100 steps): Very small rates (0.005 → 0.001) allow fine-tuning, ensuring convergence without overshooting.
- Abrupt drops: Unlike exponential or polynomial decay, the learning rate here changes suddenly at boundaries, creating a staircase effect


#### Performance scheduling

Performance-based learning rate scheduling **adjusts the learning rate dynamically** depending on the model’s validation performance (e.g., **loss or accuracy**). Mathematically, the **learning rate is reduced when the monitored metric stops improving for a certain number of steps (patience)**.



Example: If validation loss hasn’t improved for 5 epochs, reduce LR by multiplying with $\gamma$.

Qus: how we can decided which learning rate should be select in which situations?

- Power Scheduling 
- Exponential Learning Rate Scheduling 
- Piecewise constant scheduling
- Performance-based learning rate scheduling

Power scheduling is best when you want a smooth but flexible decay tied to training progress, Exponential scheduling when you need continuous, predictable reduction, and Piecewise constant scheduling when you want explicit control over phases with abrupt drops.

-  Power Scheduling (Polynomial Decay) : Large-scale models (transformers, deep CNNs) where late fine-tuning is critical.
-  Exponential Scheduling : Image classification or NLP tasks where gradual reduction stabilizes training.
- Piecewise Constant Scheduling: ResNet training on ImageNet often uses step decay (drop LR every 30 epochs).

Risks & Trade-offs
- Too aggressive decay: LR vanishes too early → underfitting.
- Too slow decay: LR stays high → oscillations, poor convergence.
- Abrupt drops (piecewise): Can cause instability unless optimizer momentum is tuned.

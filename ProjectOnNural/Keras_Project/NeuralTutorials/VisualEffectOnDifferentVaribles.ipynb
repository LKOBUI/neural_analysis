#### Apply Convolutions on Digit Mnist

This is only for purpose to see there visual effect of different filter & kernel size.

- Creates a 2D convolutional layer.
- filters=2 → The layer will learn 2 different filters (kernels). Each filter produces one feature map.
- kernel_size=7 → Each filter is a 7×7 sliding window applied to the input image

In thi API Default size:

- stride = (1,1) → filter moves one pixel at a time.
- padding = "valid" → no padding, so output shrinks compared to input.
- activation = None → raw convolution output (linear).

```python
tf.random.set_seed(42)  # extra code – ensures reproducibility
conv_layer = Conv2D(filters=2, kernel_size=7,
    strides=(1, 1),
    padding="valid",
    data_format=None,
    dilation_rate=(1, 1),
    groups=1,
    activation=None,
    use_bias=True,
    kernel_initializer="glorot_uniform",
    bias_initializer="zeros",
    kernel_regularizer=None,
    bias_regularizer=None,
    activity_regularizer=None,
    kernel_constraint=None,
    bias_constraint=None,)
feature_maps = conv_layer(images)
feature_maps.shape
```

```python
print("Feature maps shape:", feature_maps.shape)  # (1, 126, 126, num_feature)

# 4. Plot all num_feature feature maps
fig, axes = plt.subplots(1, 2, figsize=(15,6))
#for cMap in Colormap:
for i in range(2):  # Adjusted to match the number of subplots (2)
    ax = axes[i]  # Access axes as a 1D array
    ax.imshow(feature_maps[0, :, :, i], cmap='gray')  # 👉 cmap is the visual translator between numbers and colors
    ax.set_title(f"Feature {i+1}")
    ax.axis("off")
plt.tight_layout()
plt.show()
```

#### See effect of edge-detection kernel filter

```python

digit_image = digit_first

# 3. Preprocess: reshape to 4D tensor [batch, height, width, channels]
digit_input = digit_image.reshape(1, 28, 28, 1).astype(np.float32) / 255.0

# 4. Define convolution filter (kernel size 3x3, 1 input channel, 1 output channel)
# Example: simple edge-detection kernel
filter_weights = np.array([
    [[[-1]], [[0]], [[1]]],
    [[[-2]], [[0]], [[2]]],
    [[[-1]], [[0]], [[1]]]
], dtype=np.float32)

# Reshape to [filter_height, filter_width, in_channels, out_channels]
filter_weights = filter_weights.reshape(3, 3, 1, 1)

# 5. Bias = 0
bias = tf.constant([0.0], dtype=tf.float32)

# 6. Apply convolution
conv_output = tf.nn.conv2d(
    input=digit_input,
    filters=filter_weights,
    strides=[1, 1, 1, 1],
    padding="VALID"
)

# Add bias
conv_output = tf.nn.bias_add(conv_output, bias)

# 7. Visualize original digit and feature map
plt.subplot(1,2,1)
plt.imshow(digit_image, cmap='gray')
plt.title("Original Digit 5")
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(conv_output.numpy()[0,:,:,0], cmap='gray')
plt.title("Feature Map (3x3 filter)")
plt.axis('off')

plt.show()
```

#### See effect of Horizontal and vertical filters

```python
digit_image = digit_first
# 3. Preprocess: reshape to 4D tensor [batch, height, width, channels]
digit_input = digit_image.reshape(1, 28, 28, 1).astype(np.float32) / 255.0

# 4. Define horizontal and vertical filters (3x3 kernels)
# Horizontal edge detection kernel
horizontal_filter = np.array([
    [-1, -2, -1],
    [ 0,  0,  0],
    [ 1,  2,  1]
], dtype=np.float32).reshape(3, 3, 1, 1)

# Vertical edge detection kernel
vertical_filter = np.array([
    [-1, 0, 1],
    [-2, 0, 2],
    [-1, 0, 1]
], dtype=np.float32).reshape(3, 3, 1, 1)

# 5. Apply convolution with tf.nn.conv2d (bias = 0, stride=1, padding="SAME")
horizontal_output = tf.nn.conv2d(
    input=digit_input,
    filters=horizontal_filter,
    strides=[1, 1, 1, 1],
    padding="SAME"
)

vertical_output = tf.nn.conv2d(
    input=digit_input,
    filters=vertical_filter,
    strides=[1, 1, 1, 1],
    padding="SAME"
)

# 6. Plot results
plt.figure(figsize=(12,4))

plt.subplot(1,3,1)
plt.imshow(digit_image, cmap='gray')
plt.title("Original Digit 5")
plt.axis('off')

plt.subplot(1,3,2)
plt.imshow(horizontal_output.numpy()[0,:,:,0], cmap='gray')
plt.title("Horizontal Filter")
plt.axis('off')

plt.subplot(1,3,3)
plt.imshow(vertical_output.numpy()[0,:,:,0], cmap='gray')
plt.title("Vertical Filter")
plt.axis('off')

plt.show()
```

#### Qus: what is In-Channels (Input Channels) & Out-Channels (Output Channels) and where it's uses?

**In-Channels (Input Channels)**

- Definition: The number of channels in the input tensor.
- Example:
- A grayscale image has $1$ input channel.
- An RGB image has $3$ input channels.
- A feature map from a previous convolution layer might have $C channels$ (e.g., 64).

**Out-Channels (Output Channels)**
- Definition: The **number of filters applied in the convolution layer**. Each filter produces one output feature map.
- Example:
- If you define 32 filters, you get 32 output channels (feature maps).
- Each output channel corresponds to a **different learned pattern** (*edges, textures, shapes,* etc.).
=========================================================================
- *In-channels = how many “layers of information” the input has.*
- *Out-channels = how many “patterns” the network learns at this stage.*


#### Filter Tensor Shape
A convolution filter has shape:

[$f_h$ , $f_w$ , $in\_channels$ ,  $out\_channels$]

- $f_h$, $f_w$: filter height and width (e.g., 3×3)
- in_channels: matches the input depth (e.g., 3 for RGB)
- out_channels: number of filters (e.g., 64)

Another way of filter explanations:

Filter (Kernel) Tensor:
- Shape: [$f_h$,$f_w$,$C$,$K$]
- $f_h$,$f_w$: filter height and width (receptive field size)
- $C$: must match input channels
- $K$: number of output feature maps (filters applied



#### Another example:



#### Input (In-Channels): Imagine a small RGB image (10×10×3).
- Channel 0 = Red plane
- Channel 1 = Green plane
- Channel 2 = Blue plane
So the input depth = 3 in-channels.

#### Filter (Kernel): Each filter has shape [$3$,$3$, $in\_channels$,$out\_channels$].
- That means a single filter spans all 3 input channels.
- It learns a pattern across R, G, and B simultaneously.

#### Out-Channels: If you define K filters, you get K output channels.
- Example: 5 filters → output feature map depth = 5.
- Each output channel highlights a different learned feature (edges, textures, shapes, etc.).





#### Qus: Explain Padding in CNN & see the visual effect of padding on image?

In Conv2D() API:
padding: string, either "valid" or "same" (case-insensitive). 

1. If set to "$VALID$", the convolutional layer does not use zero padding, and may ignore some rows and columns at the bottom and right of the input image, depending on the stride.

    - 1.1. Qus: If we are using zero or no padding in CNN, then how ignore some rows and columns at the bottom and right of the input image ?

    Formula for output size:

    

     - Why rows/columns get ignored:
        - When you slide a convolution filter (kernel) across the input:
            - The filter must fully overlap with the input pixels to compute a valid dot product.
            - If the filter "hangs off" the edge (bottom or right), those positions are discarded because there are no pixels outside the boundary to multiply with.
            - This is why the output shrinks when you use no padding.
    Example:
    
    
    👉 The last 2 rows and 2 columns of the input are ignored because the kernel cannot fully fit there.
    - See belo code

2. If set to "$SAME$", the *convolutional layer uses zero padding if necessary*. In this case, the number of output neurons is equal to the number of input neurons divided by the stride, rounded up (in this example, 13 / 5 = 2.6, rounded up to 3). Then zeros are added as evenly as possible around the inputs.

#### Example Input Output shape analysis with No padding 

```python
from tensorflow.keras.layers import Conv2D, Input

# Example input: 5x5 grayscale image (batch size = 1)
input_data = np.arange(25).reshape((1, 5, 5, 1)).astype('float32')
print(f'Input Shape:\n{input_data[0,:,:,0]}')
# Build CNN model with Input layer
model = Sequential([
    Input(shape=(5, 5, 1)),   # define input explicitly
    Conv2D(
        filters=1,            # number of filters
        kernel_size=(3, 3),   # 3x3 kernel
        strides=(1, 1),       # stride = 1
        padding='valid'       # no padding
    )
])
# Get weights of the first Conv2D layer
weights = model.layers[0].get_weights()

# weights[0] = kernel/filter values
# weights[1] = bias values
print("Filter shape:", weights[0].shape)
filter_val = weights[0]
print("Filter values:\n", filter_val[:,:,0,0])
print("Bias values:\n", weights[1])
# Run the convolution
output = model.predict(input_data)

print("Input shape:", input_data.shape)
print("Output shape:", output.shape)
print(f'is input shape == output shape :  {input_data.shape == output.shape}')
print("Output feature map:\n", output[0, :, :, 0])
```

##### vertical filter in cnn

Vertical and horizontal filters are used for edge detection, highlighting changes in pixel intensity across the vertical or horizontal direction. These filters are applied to the input image through a process called convolution, effectively scanning the image and identifying patterns. 

Vertical Filter:
- A vertical filter emphasizes vertical edges or lines in an image by detecting changes in pixel intensity 
  along the vertical axis. 
- It typically has a structure where the left side has high positive values and the right side has high 
  negative values (or vice versa) with a neutral center. 
- When convolved with an image, a vertical filter will produce a high output value where there are strong 
  ertical edges.

##### Horizontal filter in cnn

- A horizontal filter focuses on horizontal edges or lines in an image by detecting changes in pixel intensity 
  along the horizontal axis. 
- its structure usually has positive values at the top and negative values at the bottom (or vice versa) with a 
  neutral middle. 
- When convolved with an image, a horizontal filter will produce a high output value where there are strong 
  horizontal edges.

### Pooling Layer

Their goal is to ***subsample (i.e., shrink) the input image*** in order to ***reduce the computational load***, the memory usage, and the number of parameters (thereby limiting the risk of overfitting).

- Neuron in a pooling layer is **connected to the outputs of a limited number of neurons** in the previous layer, located within a small rectangular receptive field. 
- We must define its size, the stride, and the padding type.
- A pooling neuron **$has$ $no$ $weights$**; all it does is **aggregate the inputs** using an **aggregation function such as the max or mean**.

#### what is translation invariance ?

Translation invariance means a ***system produces the same output*** regardless of the ***spatial position of its input***. In simpler terms, if you shift an object in an image, the system will still recognize it or produce the same result as if it were in its original position. 

- **Translation**:
   In the context of images, translation refers to moving an object or pattern to a different location within the image plane. 

- **Invariance**:
   Invariance, in general, means a property that remains unchanged even when certain transformations are applied. 

- **Translation Invariance**:
   Therefore, translation invariance in image processing or computer vision means that the system's output (e.g., classification, detection) is not affected by where the object appears in the image.

Qus:  What is max pooling has some downsides ?

1. It is obviously very destructive: even with a **tiny 2 × 2 kernel** and a **stride of 2**, the output will be two times smaller in both directions (so its area will be four times smaller), simply dropping 75% of the input values. 

2. Some applications, **invariance is not desirable**, for example for **semantic segmentation**: this is the **task of classifying each pixel in an image depending on the object that pixel belongs to**: obviously, if the input image is translated by 1 pixel to the right, the output should also be translated by 1 pixel to the right. The **goal in this case is equivariance**, not invariance: a small change to the inputs should lead to a corresponding small change in the output.

#### Pooling Layer TensorFlow Implementation:

Three type:
1. max pooling layer   
2. average pooling layer
3. global average pooling layer

- Max pooling: "What’s the strongest signal here?"
	- Takes the maximum value from each patch of the feature map.
   	- For example, with a $2x2 window$ and $stride 2$, it slides across the input and outputs the maximum pixel in each region.
	- Captures the most prominent feature (strongest activation).	
	- Good at preserving edges and sharp features but ignore subtle information (like      
          texture). [TODO: I need to see visually this effect]

- Average pooling: "What’s the overall signal here?"
	- Takes the average value from each patch of the feature map
	- Same sliding window as max pooling.
	- Keeps background/texture information.
	- Less effective at highlighting strong features compared to max pooling.
- Global average pooling: "Summarize the whole feature map into one number per channel."
	- Instead of sliding windows, GAP computes the average of the entire spatial dimension for each feature map.
	- If you have a feature map of size H\times W, GAP outputs a single value per channel (average of all H\times W values).
	- Reduces each feature map to one number → directly connects to classification layers
	- Prevents overfitting, reduces parameters (no fully connected layers needed)
	- Loses spatial detail completely (only global summary remains).
- Use max pooling when you want sharp, discriminative features.
- Use average pooling when you want smoother, contextual features.
- Use global average pooling when transitioning to classification, as it directly connects feature maps to output classes.

```python
from tensorflow.keras.datasets import mnist
# 1. Load MNIST dataset
(x_train, y_train), (x_test, y_test) = mnist.load_data()

# 2. Select one sample of digit "5"
digit_index = np.where(y_test == 5)[0][0]   # first occurrence of '5'
sample = x_test[digit_index]

# 3. Preprocess: reshape to (28,28,1)
sample = sample.reshape(1,28,28,1).astype("float32") / 255.0
print(f"sample image shape: {sample.shape}")
# 4. Define horizontal and vertical filters
horizontal_filter = np.array([[1, 1, 1],
                              [0, 0, 0],
                              [-1,-1,-1]], dtype="float32")

vertical_filter   = np.array([[1, 0,-1],
                              [1, 0,-1],
                              [1, 0,-1]], dtype="float32")

# Reshape filters for Conv2D (kernel_size=3x3, input_channels=1, output_channels=1)
horizontal_filter = horizontal_filter.reshape(3,3,1,1)
vertical_filter   = vertical_filter.reshape(3,3,1,1)
print(f'horizontal_filter shape : {horizontal_filter.shape}')
print(f'vertical_filter shape   : {vertical_filter.shape}')
# 5. Build two Conv2D models (one with horizontal, one with vertical filter)
number_of_feature_map = 1 # Each filter produces one output feature map.
                          # We set feature map = 1, the output will be a single feature map.
kernel_filter_size    = 3 # - The filter is a 3×3 matrix
padding_              = 'valid'
model_h = Sequential([Input(shape=(28,28,1)),   # Explicit input layer
                     Conv2D(filters = number_of_feature_map, 
                             kernel_size = kernel_filter_size,
                             strides=(1, 1),
                             padding=padding_, # No padding
                             activation=None, # not using any activations functions
                             use_bias=False
                             )])
model_v =    Sequential([Input(shape=(28,28,1)),   # Explicit input layer
                     Conv2D(filters = number_of_feature_map, 
                             kernel_size = kernel_filter_size,
                             strides=(1, 1),
                             padding=padding_, # No padding
                             activation=None, # not using any activations functions
                             use_bias=False
                             )])

# Assign weights manually
model_h.layers[0].set_weights([horizontal_filter])
model_v.layers[0].set_weights([vertical_filter])

# 6. Apply filters
output_h = model_h.predict(sample)
output_v = model_v.predict(sample)

# 7. Plot results
plt.figure(figsize=(10,4))

plt.subplot(1,3,1)
plt.imshow(sample[0,:,:,0], cmap="gray")
plt.title("Original Digit 5")
plt.axis("off")

plt.subplot(1,3,2)
plt.imshow(output_h[0,:,:,0], cmap="gray")
plt.title("Horizontal Filter")
plt.axis("off")

plt.subplot(1,3,3)
plt.imshow(output_v[0,:,:,0], cmap="gray")
plt.title("Vertical Filter")
plt.axis("off")

plt.show()
```

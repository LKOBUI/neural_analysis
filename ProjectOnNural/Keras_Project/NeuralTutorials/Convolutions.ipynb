### Filter

##### vertical filter in cnn

Vertical and horizontal filters are used for edge detection, highlighting changes in pixel intensity across the vertical or horizontal direction. These filters are applied to the input image through a process called convolution, effectively scanning the image and identifying patterns. 

Vertical Filter:
- A vertical filter emphasizes vertical edges or lines in an image by detecting changes in pixel intensity 
  along the vertical axis. 
- It typically has a structure where the left side has high positive values and the right side has high 
  negative values (or vice versa) with a neutral center. 
- When convolved with an image, a vertical filter will produce a high output value where there are strong 
  ertical edges. 

Example: [[1, 0, -1], [1, 0, -1], [1, 0, -1]] or [[-1, 1], [-1, 1]]. 
Example 2:
[1, 0, -1],
[2, 0, -2],
[1, 0, -1]

##### Horizontal filter in cnn

- A horizontal filter focuses on horizontal edges or lines in an image by detecting changes in pixel intensity 
  along the horizontal axis. 
- its structure usually has positive values at the top and negative values at the bottom (or vice versa) with a 
  neutral middle. 
- When convolved with an image, a horizontal filter will produce a high output value where there are strong 
  horizontal edges. 

##### Applications:

      - Edge Detection
      - Feature Extraction
      - Object Recognition

#### Stacking Multiple Feature Maps

At this we have stuides, output of each convolutional layer as a thin 2D layer, but in reality a convolutional layer has multiple filters.

it outputs one feature map per filter, so it is more accurately represented in 3D



```python
img_china = load_sample_image("china.jpg")/255.0
img_flower = load_sample_image("flower.jpg")/255.0
images = np.array([img_china, img_flower])
batch_size, height, width, channels = images.shape
print("Image China shape: ", images.shape)
#Create 2 filters
filters = np.zeros(shape=(7, 7, channels, 2), dtype=np.float32)
filters[:, 3, :, 0] = 1 # vertical line
filters[3, :, :, 1] = 1 # horizontal line
outputs = tf.nn.conv2d(images, filters, strides=1, padding="SAME")
plt.imshow(outputs[0, :, :, 1], cmap="gray") # plot 1st image's 2nd feature map
plt.show()

```

### Pooling Layer

Their goal is to ***subsample (i.e., shrink) the input image*** in order to ***reduce the computational load***, the memory usage, and the number of parameters (thereby limiting the risk of overfitting).





##### Size of Pooling Layer



##### Fact about Pooling Layer

pooling layer is connected to the outputs of a limited number of neurons in the previous layer, located within a small rectangular receptive field. define its size, the stride, and the padding type

A ***pooling neuron has no weights***; all it does is ***aggregate the inputs using an aggregation function*** such as the ***max or mean***.



##### what is translation invariance ?

Translation invariance means a ***system produces the same outpu***t regardless of the ***spatial position of its input***. In simpler terms, if you shift an object in an image, the system will still recognize it or produce the same result as if it were in its original position. 

- **Translation**:
   In the context of images, translation refers to moving an object or pattern to a different location within the image plane. 

- **Invariance**:
   Invariance, in general, means a property that remains unchanged even when certain transformations are applied. 

- **Translation Invariance**:
   Therefore, translation invariance in image processing or computer vision means that the system's output (e.g., classification, detection) is not affected by where the object appears in the image. 

**How it's achieved in Convolutional Neural Networks (CNNs)**: 

**Convolutional Layers**:

CNNs use convolutional layers, which apply filters (small patterns) that scan the image. These filters detect specific features (like edges, textures) and their location within the image is not crucial. The output of the convolution is translation invariant. 

**Pooling Layers**:

Pooling layers (like Max Pooling) further reduce the impact of translation by summarizing the features from a region, making the network more robust to shifts. 

**Anti-aliasing techniques**:

Techniques like blurring the image before downsampling can improve translation equivariance, which is a related but different concept, according to a blog post on Medium.

```python
early_stop_val_loss = EarlyStopping(
    monitor='val_loss',
    patience=5,
    restore_best_weights=True
)
early_stop_val_acu = EarlyStopping(
    monitor='val_Accuracy',
    patience=2,
    mode='max',
    restore_best_weights=True
)
class AccuracyThreshold(Callback):
    def __init__(self, threshold=0.98):
        super().__init__()
        self.threshold = threshold

    def on_epoch_end(self, epoch, logs=None):
        logs = logs or {}
        acc = (
            logs.get('accuracy') or
            logs.get('acc') or
            logs.get('categorical_accuracy') or
            logs.get('sparse_categorical_accuracy')
        )

        if acc is not None and acc >= self.threshold:
            print(f"\nReached {acc:.2%} accuracy, stopping training.")
            self.model.stop_training = True

class ValAccuracyThreshold(Callback):
    def __init__(self, threshold=0.98):
        super().__init__()
        self.threshold = threshold

    def on_epoch_end(self, epoch, logs=None):
        logs = logs or {}
        val_acc = logs.get('val_accuracy') or logs.get('val_acc')

        if val_acc is not None and val_acc >= self.threshold:
            print(f"\nReached {val_acc:.2%} validation accuracy, stopping training.")
            self.model.stop_training = True


callback_acu_97_percent = AccuracyThreshold(threshold=0.97)
callback_val_97_percent = ValAccuracyThreshold(threshold=0.97)
```

```python
from tensorflow.keras.datasets import mnist
(X_train, y_train), (X_test, y_test) = mnist.load_data()

# Convert & reshaping to float32
#X_train = X_train.reshape(-1, 784).astype("float32")
#X_test  = X_test.reshape(-1, 784).astype("float32")
print(f'X_train shape {X_train.shape}')
print(f'X_train shape {X_test.shape}')
X_train = X_train.reshape(-1, 28*28)
X_test  = X_test.reshape(-1, 28*28)
print(f'after reshaping X_train shape {X_train.shape}')
print(f'after reshaping X_train shape {X_test.shape}')

y_train_cat = to_categorical(y_train, 10)
y_test_cat  = to_categorical(y_test, 10)

print(f'X_train shape {X_train.shape}')
print(f'y_train shape {y_train.shape}')
print(f'X_train shape {X_test.shape}')
print(f'y_test  shape {y_test.shape}')
print(f'y_train_cat  shape {y_train_cat.shape}')
print(f'y_test_cat  shape {y_test_cat.shape}')
```

```python
# Input normalization
# Create layer
norm = Normalization()
norm.adapt(X_train)  # Prepare the normalization layer
inputs = layers.Input(shape=(28, 28,1))
InData = norm(inputs)  # Apply normalization to the input tensor
InData = Flatten()(InData)  # Flatten the normalized input
print(f'after flattern Input shape: {InData.shape}')
```

```python
layer_1 = layers.Dense(
    units = 256,  # number of neurons in hidden layer 1
    name = 'HiddenLayer-1',
    activation='relu',
    use_bias=True,
    kernel_initializer="glorot_uniform",
    bias_initializer=initializers.Zeros(),
    kernel_regularizer=None,
    bias_regularizer=None,
    activity_regularizer=None,
    kernel_constraint=None,
    bias_constraint=None,    
)(InData)
```

```python
layer_1 = BatchNormalization()(layer_1)
```

```python
layer_2 = layers.Dense(
    units = 128,  # number of neurons in hidden layer 1
    name = 'HiddenLayer-2',
    activation='relu',
    use_bias=True,
    kernel_initializer="glorot_uniform",
    bias_initializer=initializers.Zeros(),
    kernel_regularizer=None,
    bias_regularizer=None,
    activity_regularizer=None,
    kernel_constraint=None,
    bias_constraint=None,      
)(layer_1)
```

#### apply batch normalizations in layer -2

```python
layer_2 = BatchNormalization()(layer_2)
```

```python
layer_3 = layers.Dense(
    units = 64,  # number of neurons in hidden layer 1
    name = 'HiddenLayer-3',
    activation='relu',
    use_bias=True,
    kernel_initializer="glorot_uniform",
    bias_initializer=initializers.Zeros(),
    kernel_regularizer=None,
    bias_regularizer=None,
    activity_regularizer=None,
    kernel_constraint=None,
    bias_constraint=None,      
)(layer_2)
```

#### apply batch normalizations in layer -3

```python
layer_3 = BatchNormalization()(layer_3)
```

```python
layer_out = layers.Dense(
    units = 10,  # number of neurons in hidden layer 1
    name = 'HiddenLayer-out',
    activation='softmax', # output layer for multi-class classification 
    use_bias=True,
    kernel_initializer="glorot_uniform",#glorot_uniform
    bias_initializer=initializers.Zeros(),
    kernel_regularizer=None,
    bias_regularizer=None,
    activity_regularizer=None,
    kernel_constraint=None,
    bias_constraint=None,   
)(layer_3)
```

#### **Note: Not batch normalizations applyed on output layer**

```python
model = Model(inputs = InData,outputs = layer_out)
model.summary()
```

```python
run_eagerly_status = True
model.compile(
    optimizer = 'Adam',
    # ================ Loss & Loss weight ==========================================
    loss         = 'CategoricalCrossentropy',
    loss_weights = None,# assign nothing for loss weight value
    # ==============================================================================
    metrics=['Accuracy'],
    weighted_metrics=None,
    # ==============================================================================
    run_eagerly = run_eagerly_status, # how Keras executes your model internally    
    jit_compile = np.where(run_eagerly_status == True, False, True), #
    # ==============================================================================
    steps_per_execution = 1,
    auto_scale_loss = True, # Mixed precision training uses float16 or bfloat16
)
```

```python
batch_size_  = 64
epochs_      = 20
sample_size_ = X_train.shape[0]
Validation_sample_set = X_test.shape[0]
validation_batch_size_ = batch_size_
validation_freq_ = 1
print(f'Step per epoch: {math.floor(sample_size_/batch_size_) + 1}')
print(f'validation_steps: {math.floor(Validation_sample_set/validation_batch_size_) + 1}')

# Reshape the input data to match the expected input shape of the model
#X_train_flattened = X_train.reshape(-1, 784)
#X_test_flattened = X_test.reshape(-1, 784)

history_val_loss = model.fit(x = X_train, y = y_train_cat, 
                    batch_size = batch_size_,
                    epochs=epochs_,
                    verbose = 1,
                    callbacks = [early_stop_val_loss], #importent for expert user
                    validation_split = 0.0,
                    validation_data  = (X_test, y_test_cat),
                    shuffle=True,
                    class_weight  = None, # VIMP used for classification problems
                    sample_weight = None, # VIMP
                    steps_per_epoch = math.floor(sample_size_/batch_size_)+1,
                    validation_steps = math.floor(Validation_sample_set/validation_batch_size_)+1, #validation_steps_[0],                   
                    validation_batch_size = validation_batch_size_,
                    validation_freq = validation_freq_  ,                 
)
```

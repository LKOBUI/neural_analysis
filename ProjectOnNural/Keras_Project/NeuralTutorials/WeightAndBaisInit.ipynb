#### How to select Input Shape

Input shape dependent on "How many number of feature in your data set". Let dataset Delivery time data. This data set have onle two feature i.e. ***NumCase_X1, Dist_X2***  and one output ***DlvrTime_Y***.

So number of feature = 2
Number of neuron for 1st inpult layer = 



```python
X = np.array([-1, -5, 8, 3, 5]).reshape(5, 1)
y = X / 10
IN = Input(shape=(X.shape[1],))
# kernal initilazion by 1
arr = np.array([1,2]).reshape(1, 2)
val = initializers.Constant(value=arr)
VL = val(shape=(1, 2))

h1 = Dense(units = 2,kernel_initializer=VL,activation = 'sigmoid',name='dense_1')(IN)
arr = np.array([1,3]).reshape(2, 1)
val = initializers.Constant(value=arr)
VL = val(shape=(2, 1))
OUT = Dense(units = 1,kernel_initializer=VL,activation = 'relu',name='dense_2')(h1)

model = Model(IN, OUT)

# Create another model that outputs the intermediate layer
intermediate_layer_model = Model(
    inputs=model.input, outputs=model.get_layer('dense_1').output)

hidden1 = model.layers[1]
w,b = hidden1.get_weights()
print("Weights:", w,w.shape)

hidden1 = model.layers[2]
w,b = hidden1.get_weights()
print("Weights:", w,w.shape)
```

```python
#intermediate_layer_model.summary()
intermediate_output = intermediate_layer_model.predict(X)
print("Intermediate Layer Output:\n", intermediate_output)
```

```python
model.compile(loss="sparse_categorical_crossentropy",
              optimizer="sgd",
              metrics=["accuracy"])
history = model.fit(X, y, epochs=10,verbose=0)
```

### 2D feature example

```python
#Create numpy array 1 to 18 1D
X = np.random.randint(-1, 19,18).reshape(9, 2)
y = X[:, 0] + X[:, 1]
```

```python
IN = Input(shape=(X.shape[1],))

# kernal weight initilazion by 1
# weight shape = (f,u), where f = number of feature, u = number of neuron
arr = np.array([1,2,1,2,1,2]).reshape(2, 3)
val = initializers.Constant(value=arr)
VL = val(shape=(2, 3))
#initilazion of baise, 
# Bias is the linear array of shape (u,)
# where u = number of neuron in the layer
bias_arr = np.array([1,2,1])
BS = initializers.Constant(value=bias_arr)

H_1 = Dense(units = 3,kernel_initializer=VL,bias_initializer=BS,activation = 'sigmoid',name='dense_1')(IN)

# for next layer f = 3, u = 1
arr = np.array([1,3,4]).reshape(3, 1)
val = initializers.Constant(value=arr)
VL = val(shape=(3, 1))

bias_arr = np.array([5])
BS = initializers.Constant(value=bias_arr)

OUT = Dense(units = 1,kernel_initializer=VL,bias_initializer=BS,activation = 'sigmoid',name='dense_2')(H_1)

model = Model(IN, OUT)
```

```python
model.summary()
# print layer
print(model.layers[0])
print(model.layers[1])
print(model.layers[2])
```

```python
# Get H_1 & OUT weights
hidden1 = model.layers[1]
w,b = hidden1.get_weights()
print("Weights of hidden 1 layer:\n", w,"and weight shape: \n",w.shape)
print("Bias of hidden 1 layer :\n",b,"\nBias Shape:\n",b.shape)
out_layer = model.layers[2]
w,b = out_layer.get_weights()
print("Weights of Outpu layer:\n", w,"and weight shape: \n",w.shape)
print("Bias of Outpu layer :\n",b,"\nBias Shape:\n",b.shape)
```

### Weight and bais INIT for n feature and s samples

```python
# Create Numpy array m rows,n columns, and s sampls
def create_numpy_array(m, n):
    """
    Create a numpy array with m rows, n columns, and s samples.
    """
    return np.random.randint(0, 10, (m, n))
X = create_numpy_array(30, 5)
y = X[:, 0] + X[:, 1]
X.shape, y.shape
```

```python
IN = Input(shape=(X.shape[1],))
f = X.shape[1]  # number of features
u = 10 # number of neurons in the layer
# kernal weight initilazion by 1
# weight shape = (f,u), where f = number of feature, u = number of neuron
# 
arr = create_numpy_array(f, u)
val = initializers.Constant(value=arr)
VL = val(shape=(f, u))
#initilazion of baise, 
# Bias is the linear array of shape (u,)
# where u = number of neuron in the layer
bias_arr = np.array([1,2,1,2,1,2,1,2,1,2])
BS = initializers.Constant(value=bias_arr)

H_1 = Dense(units = u,kernel_initializer=VL,bias_initializer=BS,activation = 'sigmoid',name='dense_1')(IN)

f = u
u = 3

arr = create_numpy_array(f, u)
val = initializers.Constant(value=arr)
VL = val(shape=(f, u))
#initilazion of baise, 
# Bias is the linear array of shape (u,)
# where u = number of neuron in the layer
bias_arr = np.array([1,2,1])
BS = initializers.Constant(value=bias_arr)

H_2 = Dense(units = u,kernel_initializer=VL,bias_initializer=BS,activation = 'sigmoid',name='dense_2')(H_1)

f = u
u = 1

arr = create_numpy_array(f, u)
val = initializers.Constant(value=arr)
VL = val(shape=(f, u))
#initilazion of baise, 
# Bias is the linear array of shape (u,)
# where u = number of neuron in the layer
bias_arr = np.array([7])
BS = initializers.Constant(value=bias_arr)

OUT = Dense(units = u,kernel_initializer=VL,bias_initializer=BS,activation = 'sigmoid',name='dense_3')(H_2)
```

```python
model = Model(IN, OUT)
model.summary()
```

```python
# Get H_1 & OUT weights, Remember @0 always Inputlayer
hidden1 = model.layers[1]
w,b = hidden1.get_weights()
print("Weights of hidden 1 layer:\n", w,"and weight shape: \n",w.shape)
print("Bias of hidden 1 layer :\n",b,"\nBias Shape:\n",b.shape)

hidden2 = model.layers[2]
w,b = hidden2.get_weights()
print("Weights of hidden 2 layer:\n", w,"and weight shape: \n",w.shape)
print("Bias of hidden 2 layer :\n",b,"\nBias Shape:\n",b.shape)

out_layer = model.layers[3]
w,b = out_layer.get_weights()
print("Weights of Outpu layer:\n", w,"and weight shape: \n",w.shape)
print("Bias of Outpu layer :\n",b,"\nBias Shape:\n",b.shape)

```

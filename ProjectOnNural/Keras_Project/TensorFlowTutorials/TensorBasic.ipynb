#### Ragged tensors (non-uniform shapes):

Ragged tensors are a specialized type of tensor designed to efficiently handle data with non-uniform shapes, particularly when dealing with variable-length sequences or hierarchical data structures. Unlike standard tensors which require all elements along a given dimension to have the same length (resulting in a rectangular or fixed-size structure), ragged tensors allow for varying lengths in one or more of their dimensions.

#### Sparse tensors

A sparse tensor is a data structure, specifically a multidimensional array, ***where most of the elements are zero*** &
***reduces storage requirements*** and ***computational load*** associated with ***unnecessary zero values***.

Instead of storing all the elements, ***sparse tensors store only the non-zero values along with their corresponding indices***, making them memory-efficient for datasets with ***many zero entries***.

This approach is particularly useful in scenarios like natural language processing (NLP) and computer vision where data often exhibits sparsity. 

- High dimensionality: Tensors are multidimensional arrays, extending the concept of matrices to higher dimensions.
- Sparsity: A significant portion of the tensor's elements are zero.
- Efficiency: Sparse tensors store only the non-zero values and their locations (indices), reducing storage space and 
              computational overhead.

Common use cases:

- NLP: Representing word embeddings, TF-IDF vectors, etc. 
- Computer vision: Processing images with many dark pixels (zero values). 
- Graph processing: Representing sparse graph structures. 
- Large-scale data analysis: Handling datasets with inherent sparsity.

### Custom Metrics

Losses and metrics are conceptually not the same thing. 

A metric is a function that is used to judge the performance of your model. 

Metric functions are similar to loss functions, except that the results from evaluating a metric are not used when training the model. Note that you may use any loss function as a metric.

##### Losses:

losses are used by Gradient Descent to train a model, so they must be differentiable (at least where they are 
evaluated) and their gradients should not be 0 everywhere.

##### Metrics:

metrics are used to evaluate a model, they must be more easily interpretable, and they can be nondifferentiable
or have 0 gradients everywhere (e.g., accuracy).

***Note:*** Most cases, defining a custom metric function is exactly the same as defining a custom loss function.So we could even use the Huber loss function as a metric

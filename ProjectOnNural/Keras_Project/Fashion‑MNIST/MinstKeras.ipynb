    - The activity_regularizer applies a penalty to those activations during training, by adding an extra

    - To control the magnitude of activations (prevent them from growing too large).
    - To improve generalization and reduce overfitting

   - Autoencoders → encourage sparse hidden representations.
   - Feature selection → L1 activity regularization can force only a few neurons to be active.
   - Preventing over-activation → keeps outputs stable and avoids exploding values.

   - 3.3.1 The bias_regularizer parameter lets you apply a regularization function to those bias values during training.

 - 4.1. Match initializer to activation function   
   

5. What is bias_initializer & bias_constraint ? How they are diffrent ?
- bias_initializer = sets the starting values of biases.
- bias_constraint = enforces rules on biases during training.
  

6.  How to Select Shape in Keras
     - Kernel shape = (input_dim , units)
        - input_dim = number of regressors (independent variables)
        - unit → number of neurons you specify in the Dense layer.
        - Input has 100 features, Dense layer has 64 units
        - dense = layers.Dense(64, kernel_initializer='he_normal')

        - layers.Dense(1, activation='sigmoid')              # target

7. What is kernel_constraint ? [kernel_constraint for the main weights matrix]
   - Unlike kernel_regularizer (which adds a penalty to the loss), constraints directly restrict the values of weights to satisfy certain conditions.
   - In a Dense layer, the kernel is the weight matrix connecting inputs to outputs.
   - The kernel_constraint parameter lets you apply a constraint function to those weights after each update during training.
   - bias_constraint for the bias


### initialize a Dense layer in Keras

#### Layer - 1

```python
# =============================== Layer name =================================================
Layer_name = 'Layer_1'
# =============================== UINTS ======================================================
number_of_neurons          = 300 #parameter specifies the number of neurons (nodes) in that fully connected (Dense) layer.

INPUT_DIM_SIZE = InData.shape # total numeber of regressor variable ni layer 1
# =============================== activation_functions =======================================
activation_functions = ['relu','sigmoid','softmax','tanh','linear']
# =============================== use_bias ===================================================
USE_BIAS             = True, 
# if use_bias = False; then output = activations((input @ kernel))
# if use_bias = True; then output = activations((input @ kernel) + bias)
# =============================== kernel & base initializer ==================================
kernel_initializer_list = ['glorot_uniform','he_normal',\
                           'he_uniform','random_norm',\
                           'random_unifc','zeros','ones']
# - parameter controls how those bias values are initialized at the start of training.
bias_initializer_zeros  = initializers.Zeros()
bias_initializer_const  = initializers.Constant(0.1)
bias_initializer_random = initializers.RandomNormal(mean=0.0, stddev=0.05)                            
# =============================== Set initializer and activation function ===================
# ReLU / variants (LeakyReLU, ELU) -> he_normal or he_uniform
# Sigmoid / Tanh                   -> glorot_uniform
# Linear / Softmax                 -> glorot_uniform,random_norm
# Custom activations               -> VarianceScaling
activation_initializer_function = ['relu','he_normal']
# =============================== kernel & base regularizers =================================
kernel_regularizer_l2       = regularizers.l2(0.01)   # λ = 0.01 [Ridge Regressions]
kernel_regularizer_l1       = regularizers.l1(0.001)  # λ = 0.001 [Lasso Regressions]
kernel_regularizer_l1_l2    = regularizers.l1_l2(l1=0.001, l2=0.01) # Combination of both
kernel_regularizer_default  = None
bias_regularizer_l2         = regularizers.l2(0.01)   # λ = 0.01
bias_regularizer_l1         = regularizers.l1(0.001)  # λ = 0.001
bias_regularizer_l1_l2      = regularizers.l1_l2(l1=0.001, l2=0.01)
bias_regularizer_default    = None
# =============================== activity_regularizer =======================================
# The activity_regularizer applies a penalty to those activations during training, 
# by adding an extra term to the loss function.
# Why use it? -> - control the magnitude of activations (prevent them from growing too large).
activity_regularizer_l1      = regularizers.l1(0.01)   # λ = 0.01
activity_regularizer_l2      = regularizers.l2(0.01)
activity_regularizer_l1_l2   = regularizers.l1_l2(l1=0.001, l2=0.01)
activity_regularizer_dafault = None
# =============================== kernel bias constraint ==========================================
kernel_constraint_max_norm  = keras.constraints.MaxNorm(max_value=2.0),# See Keras for more argument
kernel_constraint_NonNeg    = keras.constraints.NonNeg(),
kernel_constraint_UnitNorm  = keras.constraints.UnitNorm(),
kernel_constraint_default   = None 

bias_constraint_MAxNorm     = keras.constraints.MaxNorm(max_value=2.0)
bias_constraint_NonNeg      = keras.constraints.NonNeg()
bias_constraint_MinMaxNorm  = keras.constraints.MinMaxNorm(min_value=0.0, max_value=1.0, rate=1.0)
bias_constraint_default     = None     # applied to the bias vector instead of the weight matrix.

lora_rank_default           = None
lora_alpha_default          = None
```

#### A.2 Losses

Three type od losses are define

- Probabilistic losses
- Regression losses
- Hinge losses for "maximum-margin" classification

**Loss 1: Probabilistic loss functions and when to use them:**

1. Binary Classification (Two Classes: 0 or 1, Yes/No)
  - BinaryCrossentropy
  - sigmoid (Activation)
2. Multi-Class Classification (More than Two Mutually Exclusive Classes)
  - CategoricalCrossentropy
    - softmax 
    - Use when your true labels are one-hot encoded 
  - SparseCategoricalCrossentropy
    - softmax 
    - Use when your true labels are integers (e.g., 1, 2, 0, 1).
3. Multi-Label Classification (Multiple Classes Simultaneously) 
  - BinaryCrossentropy
  - sigmoid 
  - The standard approach is to treat each possible label as a separate binary classification problem.
4. Specialized Probabilistic Losses:
  - KLDivergence (Kullback-Leibler Divergence): 
    - Measures the difference between two probability distributions. 
    - It is useful in applications where you are learning a complex distribution, such as in Variational Autoencoders (VAEs).
  - FocalCrossentropy (or BinaryFocalCrossentropy, CategoricalFocalCrossentropy): 
    - An extension of cross-entropy used to address severe class imbalance problems, where easy examples are down-weighted to focus training on hard-to-classify examples.
  - Poisson Loss: 
    - Used for count data problems where the target variable is a non-negative integer (e.g., number of customers visiting a website)

#### A.3 Metrics:
A metric is a function that is used to **judge the performance of your model**.

Metric functions are similar to loss functions, except that the results from evaluating a metric are not used when training the model. Note that you may use any loss function as a metric.

- Available metrics
  - Base Metric class
  - Accuracy metrics
  - Probabilistic metrics
  - Regression metrics
  - Classification metrics based on True/False positives & negatives
  - Image segmentation metrics
  - Hinge metrics for "maximum-margin" classification
  - Metric wrappers and reduction metrics

####  explanation of which metrics to use and why ?

***Conditions When to Use Metrics***
- Classification (balanced data) → accuracy
- Classification (imbalanced data) → AUC, precision, recall, F1Score
- Binary classification → binary_accuracy, precision, recall
- Multi-class classification → categorical_accuracy or sparse_categorical_accuracy
- Regression → mae, mse, RootMeanSquaredError
- Ranking / recommendation → TopKCategoricalAccuracy



Key Tips
- Loss drives training, metrics are for monitoring only.
- Choose metrics that reflect your business or scientific goal (e.g., recall for medical diagnosis, precision for spam detection).
- For imbalanced datasets, accuracy can be misleading → prefer AUC, precision, recall, or F1.
- You can define custom metrics using TensorFlow/Keras functions if built-ins don’t fit

#### A.5 🌟 Why Do We Need Auto Loss Scaling ?

In Keras/TensorFlow, auto loss scaling (via the LossScaleOptimizer) is used in mixed precision training to prevent numeric underflow when working with low‑precision 

- Mixed precision training uses float16 or bfloat16 to speed up training and reduce memory usage.
- Problem: float16 has a limited range → very small gradient values can underflow to zero, breaking training.
- Solution: Loss scaling multiplies the loss by a factor before computing gradients, making them larger and less likely to vanish. Afterward, gradients are divided back to restore their correct scale.

#### ⚙️ Relationships Between Them

- Batch Size ↔ Steps per Epoch
  - **Larger batch → faster** but memory-heavy; 
  - **smaller batch → slower** but more generalizable
  - Steps per epoch = $\frac{Number\  of\  samples}{Batch \ Size}$.
  - Example: 60,000 samples, batch size 128 → ~469 steps per epoch.

- Epochs ↔ Training Duration
  - **More epochs** → **more learning**, but **risk of overfitting** 
  - **Total training iterations** = Epochs × Steps per epoch.
  - More epochs = more updates, but diminishing returns after convergence.

- Validation Steps ↔ Validation Batch Size
  - Controls how much of validation set is used each time
  - Validation steps = $\frac{Validation \ samples}{Validation \ Batch \ Size}$   
  - Example: 10,000 validation samples, batch size 64 → ~157 validation steps.

- Validation Frequency ↔ Epochs
  - If validation_freq = 1 → validation after every epoch.
  - If validation_freq = 2 → validation after every 2 epochs.
  - This reduces overhead if validation is expensive.

#### Putting It Together
Imagine you have:
- Training set: 240,000 samples
- Validation set: 80,000 samples
- Batch size: 64
- Validation batch size: 64
- Epochs: 10
- Validation frequency: 1
Then:
- Steps per epoch = 240,000 ÷ 64 = 3,750
- Validation steps = 80,000 ÷ 64 = 1,250
- Total training iterations = 10 × 3,750 = 37,500 updates
- Validation runs = 10 (once per epoch)

#### Trade-offs & Best Practices
- Batch size too large → faster but may generalize poorly.
- Too many epochs → overfitting risk.
- Validation too frequent → slows training.
- Validation too infrequent → may miss early signs of overfitting.
#### Rule of thumb:
- Batch size = 32 or 64 for balance.
- Epochs = tune based on convergence.
- Validation frequency = 1 (every epoch) for most cases.
- Adjust validation steps if validation set is huge to save time.

#### C.2. Learning Rate Scheduler

- Customizable: You must define the schedule function carefully; a bad schedule can destabilize training.
- Verbose=1: Useful for debugging, shows LR per epoch.
- 👉 Combine with EarlyStopping/ModelCheckpoint: Ensures you don’t overtrain while experimenting with LR schedules.


```python
# 👉 Learning rate halves every 10 epochs
def step_decay(epoch, lr):
    drop = 0.5
    epochs_drop = 10
    if epoch % epochs_drop == 0 and epoch:
        return lr * drop
    return lr

# 👉 Learning rate smoothly oscillates between max_lr and min_lr
def cosine_annealing(epoch, lr):
    max_lr = 0.01
    min_lr = 0.0001
    T_max = 50  # total epochs
    return min_lr + (max_lr - min_lr) * (1 + np.cos(np.pi * epoch / T_max)) / 2

# 👉 Gradually increases LR for first 5 epochs, then decays
def warmup_decay(epoch):
    if epoch < 5:   # warmup phase
        return 0.001 * (epoch + 1)
    else:           # decay phase
        return 0.01 * (0.1 ** ((epoch - 5) // 10))

lr_scheduler_cb_step_decay       = LearningRateScheduler(step_decay, verbose=1) 
lr_scheduler_cb_cosine_annealing = LearningRateScheduler(cosine_annealing, verbose=1)
lr_scheduler_cb_warmup_decay     = LearningRateScheduler(warmup_decay, verbose=1)
```

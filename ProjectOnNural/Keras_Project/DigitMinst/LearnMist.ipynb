```python
path = os.path.join(os.getcwd(), 'mnist_784.csv')
df = pd.read_csv(path)
print(f"total Pixal Len: {len(df.columns.values)}")
print(df.shape[0], 'Total Pixal 28x28')
```

```python
df = df.loc[0:19,:]
X = df.drop('class',axis=1)
y = df['class']
y_one_hot = to_categorical(y, num_classes=10)
print(f'x shape:{X.shape} y shape {y.shape} y_one_hot {y_one_hot.shape}')
```

```python
inputs = Input(shape=(28, 28, 1), name="image_input")
inputs = Input(shape=(784,), name="input_layer")
inputs.shape
```

```python
l1 = layers.Dense(units = 160, 
                  name = 'layer_1',
                  activation="relu",
                  use_bias=True,
                  kernel_initializer="glorot_uniform",
                  bias_initializer="zeros"
                  )(inputs)
l2 = layers.Dense(units = 40, name = 'layer_2',activation="relu",use_bias=True,kernel_initializer="glorot_uniform",
             bias_initializer="zeros")(l1)
outputs = layers.Dense(units = 10, name = 'out_layer',activation="softmax",use_bias=True,kernel_initializer="glorot_uniform",
             bias_initializer="zeros")(l2)
model = keras.Model(inputs=inputs, outputs=outputs)
```

```python
model.get_layer(index=1)
```

```python
print(f'Input shape:{inputs.shape}')
print(f'l1 shape:{l1.shape}')
print(f'l2 shape:{l2.shape}')
print(f'outputs shape:{outputs.shape}')
```

```python
# 
layer = model.get_layer("layer_1")
# Get weights
weights = layer.get_weights()
kernel, bias = weights[0], weights[1]
print("Kernel shape:", kernel.shape,'X shape: ',X.shape)  # (784, 128)
print("Bias shape:", bias.shape,type(bias))      # (128,)
print(f'Shape X@W {(X @ kernel).shape} baise shape {bias.shape}')
print(f'final shape: {((X @ kernel) + bias).shape}')
```

```python
# Access first Dense layer
layer = model.get_layer("layer_2")
# Get weights
weights = layer.get_weights()
kernel, bias = weights[0], weights[1]
print("Kernel shape:", kernel.shape)  # (784, 128)
print("Bias shape:", bias.shape)      # (128,)
```

```python
# Access first Dense layer
layer = model.get_layer("out_layer")
# Get weights
weights = layer.get_weights()
kernel, bias = weights[0], weights[1]
print("Kernel shape:", kernel.shape)  # (784, 128)
print("Bias shape:", bias.shape)      # (128,)
for i, w in enumerate(weights):
        print(f"  Weight {i} shape: {w.shape}")
        wr = np.round(w,3)[0:9]
        print(wr)   # actual matrix values

print(np.round(bias,3))
```

```python
# Get the total number of layers
total_layers = len(model.layers)
# Get the maximum valid index
max_index = total_layers - 1
print(f"Total number of layers in the model: {total_layers}")
print(f"The maximum layer index is: {max_index}")
# You can now iterate through all layers by index
print("\nLayer details:")
for i in range(total_layers):
    layer = model.get_layer(index=i)
    # 2. The layer is built, so we can immediately check initial random weights
    initial_weights = layer.get_weights()
    print(f"Number of weight tensors returned: {len(initial_weights)}")    
    print(f"Index {i} Layer Name: {layer.name}, Type: {type(layer).__name__}")
    if i > 0:
        if len(weights) > 0:
            kernel, bias = weights[0], weights[1]
            print("Kernel shape:", kernel.shape)  # (784, 128)
            print("Bias shape:", bias.shape)      # (128,)
    


```

```python
model.compile(
    optimizer = 'Adam',
    # ================ Loss & Loss weight ==========================================
    loss         = 'CategoricalCrossentropy',
    loss_weights = None,# assign nothing for loss weight value
    # ==============================================================================
    metrics=['Accuracy'],
    
)
```
